```python
import numpy as np

# ====================
# 读取数据
# ====================
data = np.load("./data/data_50k.npz")
X = data['x_train']
Y = data['y_train']

# ====================
# 随机打乱特征顺序
# ====================
n_features = X.shape[1]
shuffled_order = np.random.permutation(n_features)
X_shuffled = X[:, shuffled_order]

# ====================
# 原始 causal variants
# ====================
all_variables = np.array([
    100, 200, 300, 400, 500, 600, 700, 800, 900, 1000, 1100, 1200, 
    1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000, 2100, 2200, 
    2300, 2400, 13000,13100,13200,13300,13400,13500,13600,13700,
    13800,13900,14000,14100,14200,14300,14400,14500,14600,14700,
    14800,14900,15000,15100,15200,15300
])

# ====================
# 生成反向映射（原 index → 新 index）
# ====================
inverse_map = np.zeros(n_features, dtype=int)
inverse_map[shuffled_order] = np.arange(n_features)

# ====================
# 更新 causal 列索引
# ====================
all_variables_new = inverse_map[all_variables]

print("新的 causal indices:", all_variables_new)


```

```python
np.savez("./data/shuffled_new_data_50k.npz",
         x_train=X_shuffled,
         y_train=Y,
         shuffled_order=shuffled_order,
         causal_new=all_variables_new)
```
