# Feature Selection Commands — ALLAML
Configurations are placed at the top for easy switching of data paths and names.

- **Search 统一默认**：`--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss`
- **Evaluation 统一默认**：`--use_evaluation 1 --max_features 10 --feature_step 2`
- **is_snp**：固定 `0`
- 其它模型超参走程序默认，不在命令里显式覆盖。

```python
DATA_PATH = "./data/ALLAML_10.npz"
NAME = "ALLAML"

BASE = f"python ./main.py --input_path {DATA_PATH} --name {NAME}"

EVAL = "--use_evaluation 1 --max_features 10 --feature_step 2"
SEARCH = "--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss"
NO_EVAL = "--use_evaluation 0"
NO_SEARCH = "--do_parameter_search 0"
IS_SNP = "--is_snp 0"

```

```python
DATA_PATH = "./data/ALLAML_10.npz"
NAME = "ALLAML"

BASE = f"python ./main.py --input_path {DATA_PATH} --name {NAME}"

EVAL = "--use_evaluation 1 --max_features 10 --feature_step 2 --max_features_graces 10"
SEARCH = "--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss"
NO_EVAL = "--use_evaluation 0  --max_features_graces 10"
NO_SEARCH = "--do_parameter_search 0"
IS_SNP = "--is_snp 0"

methods = [
    "GRACES",
    "EARFS",
    "CANCELOUT",
    "STG",
    # "CAE",  # ignore CAE for now
    "DeepLIFT",
    "GradientShap",
    "FeatureAblation",
    "Occlusion",
    "Lime",
]

settings = [
    ("no_search_no_eval", NO_SEARCH, NO_EVAL),
    # ("search_no_eval", SEARCH, NO_EVAL),
    # ("no_search_eval", NO_SEARCH, EVAL),
    # ("search_eval", SEARCH, EVAL),
]

for method in methods:
    for setting_name, search_arg, eval_arg in settings:
        print("=" * 80)
        print(f"Running method={method}, setting={setting_name}")
        print("=" * 80)

        cmd = f"{BASE} --method {method} {IS_SNP} {search_arg} {eval_arg}"
        print(cmd)

        !{cmd}
```

```python
DATA_PATH = "./data/ALLAML_10.npz"
NAME = "ALLAML"

BASE = f"python ./main.py --input_path {DATA_PATH} --name {NAME}"

# CAE-specific K values
CAE_K = "--cat_k_select 1,5,10,15,20"

# Evaluation/search settings
EVAL = "--use_evaluation 1 --max_features 10 --feature_step 2"
NO_EVAL = "--use_evaluation 0"
SEARCH = "--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss"
NO_SEARCH = "--do_parameter_search 0"
IS_SNP = "--is_snp 0"

method = "CAE"

settings = [
    # ("no_search_no_eval", NO_SEARCH, NO_EVAL),
    # ("search_no_eval", SEARCH, NO_EVAL),
    # ("no_search_eval", NO_SEARCH, EVAL),
    ("search_eval", SEARCH, EVAL),
]

for setting_name, search_arg, eval_arg in settings:
    print("=" * 80)
    print(f"Running method={method}, setting={setting_name}")
    print("=" * 80)

    cmd = f"{BASE} --method {method} {IS_SNP} {CAE_K} {search_arg} {eval_arg}"
    print(cmd)

    !{cmd}
```
