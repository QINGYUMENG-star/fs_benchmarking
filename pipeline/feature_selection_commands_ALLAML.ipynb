# Feature Selection Commands — ALLAML
配置统一放在最上面，方便你快速切换数据路径和名称。

- **Search 统一默认**：`--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss`
- **Evaluation 统一默认**：`--use_evaluation 1 --max_features 10 --feature_step 2`
- **is_snp**：固定 `0`
- 其它模型超参走程序默认，不在命令里显式覆盖。

```python
# ==== 配置（按需改动） ====
DATA_PATH = "./data/ALLAML_10.npz"
NAME = "ALLAML"

# 便捷别名（不要改）
BASE = f"python ./main.py --input_path {DATA_PATH} --name {NAME}"
EVAL = "--use_evaluation 1 --max_features 10 --feature_step 2"
SEARCH = "--do_parameter_search 1 --n_trials 10 --n_jobs 4 --eval_metric loss"
NO_EVAL = "--use_evaluation 0"
NO_SEARCH = "--do_parameter_search 0"
IS_SNP = "--is_snp 0"
```

## EARFS

**A. 不搜参 / 不评估**
